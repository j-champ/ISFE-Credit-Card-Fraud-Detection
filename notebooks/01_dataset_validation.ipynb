```python
DATA_PATH = Path("../data/raw/creditcard.csv")

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully")
print("Shape:", df.shape)

df.head()
```

```python
class_counts = df["Class"].value_counts()

class_percentages = df["Class"].value_counts(normalize=True) * 100

distribution = pd.DataFrame({
    "Count": class_counts,
    "Percentage": class_percentages
})

distribution
```

```python
duplicate_groups = (
    df.groupby(list(df.columns), dropna=False)
      .size()
      .reset_index(name="frequency")
)

duplicate_groups = duplicate_groups[
    duplicate_groups["frequency"] > 1
]

print("Number of unique duplicate groups:",
      len(duplicate_groups))

print("\nDuplicate group sizes:")
print(duplicate_groups["frequency"].value_counts().sort_index())

print("\nTotal excess duplicate occurrences:",
      (duplicate_groups["frequency"] - 1).sum())

print("\nDuplicate groups by class:")
print(duplicate_groups.groupby("Class")["frequency"].agg(
    ["count", "sum"]
))
```

```python
fraud_duplicate_groups = duplicate_groups[
    duplicate_groups["Class"] == 1
]

print("Unique fraudulent duplicate groups:",
      len(fraud_duplicate_groups))

print("\nFraud duplicate frequencies:")
print(fraud_duplicate_groups["frequency"].value_counts())

print("\nFraudulent duplicate records:")
display(fraud_duplicate_groups.sort_values(
    "frequency", ascending=False
))
```
